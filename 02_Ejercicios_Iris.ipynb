## Ejercicio 3. Filtrar con una o dos condiciones

a) Selecciona solamente flores de la especie `setosa`. b) Selecciona flores de `virginica` cuyo `petal_length_cm` sea **mayor que 5.5**. c) Muestra cuántas filas tiene cada resultado.

**Pista:** Utiliza una máscara de comparación y `&`; cada condición debe ir entre paréntesis.


```python
mascara1 = iris["species"] == "setosa"
seleccion1 = iris.loc[mascara1, ["species", "petal_length_cm"]]
print("Coincidencias setosa:", len(seleccion1)) #a

mascara2 = (iris["species"] == "virginica") & (iris["petal_length_cm"] > 5.5)
seleccion2 = iris.loc[mascara2, ["species", "petal_length_cm"]]
print("Coincidencias virginica con largo de pétalo > 5.5 cm:", len(seleccion2)) #b
```

## Ejercicio 4. Contar y agrupar por especie

a) Cuenta cuántas flores hay de cada especie. b) Calcula el **promedio de `petal_length_cm` para cada especie**. c) Identifica cuál tiene el promedio mayor.

**Pista:** Busca `value_counts`, `groupby` y `mean`. La columna `petal_length_cm` no contiene faltantes en esta copia.


```python
species_counts = iris["species"].value_counts()
print("Cantidad de flores por especie:\n", species_counts) #a

groupby_species = iris.groupby("species")
mean_petal_length = groupby_species["petal_length_cm"].mean()
print("\nPromedio de largo de pétalo por especie:\n", mean_petal_length) #b

for species, mean_length in mean_petal_length.items():
    if mean_length == mean_petal_length.max():
        print(f"\nEspecie con mayor promedio de largo de pétalo: {species} ({mean_length} cm)") #c
```
